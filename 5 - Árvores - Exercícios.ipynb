1. Implemente uma classe chamada NodoArvore que representa um nodo de uma árvore.
Cada nodo deve ter:
	1.	Um atributo nodo, que armazena o nodo.
	2.	Um atributo filhos, que será uma lista para armazenar os nodos filhos.
	3.	Um método adicionar_filho(nodo_filho) que adiciona um novo nodo à lista de filhos.

* No programa principal:
	*	Crie um nodo raiz com valor 1.
	*	Adicione três nodos filhos com valores 2, 3 e 4.
  	*   Para cada nodo filho, adicione mais 2 valores de exemplo.
	*	Exiba o valor da raiz e de seus filhos.

````python
# resolva o exercício aqui
class NodoArvore:
    def __init__(self, nodo):
        self.nodo = nodo
        self.filhos = []

    def adicionar_filho(self, nodo_filho):
        self.filhos.append(nodo_filho)

    def __str__(self, level = 0):
        ret = ("   " * level) + (str(self.nodo) + "\n")
        for filho in self.filhos:
            ret += filho.__str__(level + 1)
        return ret


raiz = NodoArvore("1")
filho_2 = NodoArvore("2")
filho_3 = NodoArvore("3")
filho_4 = NodoArvore("4")
filho_2a = NodoArvore("2A")
filho_2b = NodoArvore("2B")
filho_3a = NodoArvore("3A")
filho_3b = NodoArvore("3B")
filho_4a = NodoArvore("4A")
filho_4b = NodoArvore("4B")


raiz.adicionar_filho(filho_2)
raiz.adicionar_filho(filho_3)
raiz.adicionar_filho(filho_4)

filho_2.adicionar_filho(filho_2a)
filho_2.adicionar_filho(filho_2b)

filho_3.adicionar_filho(filho_3a)
filho_3.adicionar_filho(filho_3b)

filho_4.adicionar_filho(filho_4a)
filho_4.adicionar_filho(filho_4b)

print(raiz)
````

2. Implemente uma função calcular_altura(nodo) que receba um nodo raiz e retorne a altura da árvore.
Definição:
* A altura de um nodo é o número de arestas no caminho mais longo até uma folha.
* A altura de uma folha é 0.

No programa principal, crie uma árvore manualmente (como no exercício 2) e exiba a altura da árvore.

````python
# resolva o exercício aqui

def calcular_altura(nodo):
    if not nodo.filhos:
        return 0
    else:
        return 1 + max(calcular_altura(filho) for filho in nodo.filhos)

r = calcular_altura(raiz)

print(f"A altura da arvore é: {r}")
````

3. Implemente uma função contar_nodos(nodo) que receba um nodo raiz e retorne o total de nodos na árvore.
Regras:
    1.	Se o nodo for nulo, retorne 0.
	2.	Caso contrário, retorne 1 (para contar o nodo atual) mais a soma dos nodos de todos os filhos.

No programa principal, crie uma árvore manualmente e exiba a contagem total de nodos.

````python
# resolva o exercício aqui

def contar_nodos(nodo):
    if nodo is None:
        return 0
    else:
        return 1 + sum(contar_nodos(filho) for filho in nodo.filhos)

total = contar_nodos(raiz)
print(f"O total de nodos na arvore é: {total}")
````

4. Reescreva a estrutura de uma árvore usando dicionários em vez de classes.\
Exemplo:\
A árvore:

          "A"
         / | \
      "B" "C" "D"

Pode ser representada como:

```python
arvore = {
    "A": ["B", "C", "D"],
    "B": [],
    "C": [],
    "D": []
}

Faça o seguinte:
1.	Crie uma função exibir_arvore(arvore, nodo) que exiba todos os nodos e seus filhos, começando a partir de um nodo raiz fornecido.
2.	No programa principal, defina a árvore acima como um dicionário e chame a função para exibir a estrutura da árvore.

````python
# resolva o exercício aqui
def exibir_arvore(arvore, nodo, nivel = 0):
    print("\t" * nivel + nodo)
    for filho in arvore[nodo]:
        exibir_arvore(arvore, filho, nivel + 1)

arvore = {
    "A": ["B", "C", "D"],
    "B": ["D", "E"],
    "C": [],
    "D": ["G"],
    "E": ["F"],
    "F": [],
    "G": ["H"],
    "H": [],
}

exibir_arvore(arvore, "A")
````
